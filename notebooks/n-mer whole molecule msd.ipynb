```python
import matplotlib.pyplot as plt
import numpy as np
import signac
import gsd.hoomd
```

```python
project = signac.get_project("/home/stephaniemccallu/scratch/Entanglements")
```

## 100-mer

```python
for length, jobs in project.find_jobs({"doc.sampled":True}).groupby("doc.lengths"):
    for job in jobs:
        if length == 100:    
            time = np.load(job.fn("msd_time.npy"))
            msd_nm = np.load(job.fn("msd_data_reduced.npy"))
            plt.plot(time[1:], msd_nm[1:], linewidth=3, label=length)

'''plt.plot(
    time[1:150],
    time[1:150]**(0.6) * 5e6,
    color="black",
    linestyle="--"
)

plt.plot(
    time[1:150],
    time[1:150]**(0.25) * 16e3,
    color="red",
    linestyle="--"
)
plt.plot(
    time[10:],
    time[10:] * 2e9,
    color="black",
    linestyle="--"
)
plt.text(0.7e-8, 120, "$\propto t^{0.6}$",fontsize=12, color="black")
#plt.text(0.7e-8, 250, "$\propto t^{0.25}$",fontsize=12, color="red")
plt.text(1e-7, 150, "$\propto t$",fontsize=12, color="black")
'''
plt.xscale("log")
plt.yscale("log")
plt.xlabel("time [s]")
plt.ylabel("MSD [sigma]")
plt.legend(title="N")
plt.title("MSD of original trajectory")
```
