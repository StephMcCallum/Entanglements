```python
import matplotlib.pyplot as plt
import numpy as np
import signac
import gsd.hoomd
import math
```

```python
project = signac.get_project("/home/stephaniemccallu/scratch/Entanglements")
```

## 100-mer

```python
for length, jobs in project.find_jobs({"doc.sampled":True}).groupby("doc.lengths"):
    for job in jobs:
        if length == 400:    
            time = np.load(job.fn("msd_time_bs_50_400.npy"))
            msd_nm = np.load(job.fn("msd_data_reduced_bs_50_400.npy"))
            w = 25
            smooth = np.convolve(msd_nm, np.ones(w), 'valid') / w
            smooth_time = time[w//2:-w//2+1]
            msd_first_der = np.gradient(np.log(smooth[1:]),np.log(smooth_time[1:]))
            #print(msd_first_der)
            plt.plot(smooth_time[1:], msd_first_der, linewidth=3, label="smooth")
            #plt.plot(time[1:], msd_nm[1:], linewidth=3, label=length)

'''plt.plot(
    time[1:150],
    time[1:150]**(0.6) * 4.5e6,
    color="black",
    linestyle="--"
)

plt.plot(
    time[1:150],
    time[1:150]**(0.25) * 16e3,
    color="red",
    linestyle="--"
)
plt.plot(
    time[40:],
    time[40:] * 2e9,
    color="black",
    linestyle="--"
)
plt.text(0.7e-8, 120, "$\propto t^{0.6}$",fontsize=12, color="black")
#plt.text(0.7e-8, 250, "$\propto t^{0.25}$",fontsize=12, color="red")
plt.text(5e-7, 350, "$\propto t$",fontsize=12, color="black")
'''
#plt.xscale("log")
#plt.yscale("log")
plt.xlabel("time [s]")
plt.ylabel("MSD [sigma]")
plt.legend(title="N")
plt.title("MSD gradient")
```
